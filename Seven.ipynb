```python
from sklearn.preprocessing import LabelEncoder

df=pd.read_csv('/content/drive/MyDrive//Housing.csv')
label_encoder = LabelEncoder()
y=df[['mainroad']].head()

y['mainroad_encoded'] = label_encoder.fit_transform(y['mainroad'])
print("\nAfter Label Encoding 'mainroad':")
print(y[['mainroad', 'mainroad_encoded']])


```

```python
df_encoded=pd.get_dummies(df,drop_first=True)
X = df_encoded.drop('parking', axis=1)
y = df_encoded['parking']

print("X has all house information ")
print("y has parking  ")

# Split: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,    # 20% for test
    random_state=20   # Same result every time
)
```

```python
scaler = StandardScaler()

# Learn from training data and scale it
X_train_scaled = scaler.fit_transform(X_train)

# Scale test data (just transform, don't fit)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression
# Create model
model = LogisticRegression()

# Train the model
model.fit(X_train_scaled, y_train)
print("✅ Model trained!")

# Check accuracy on training data
train_score = model.score(X_train_scaled, y_train)
print(f"Training accuracy: {train_score:.1%}")

# Check accuracy on test data
test_score = model.score(X_test_scaled, y_test)
print(f"Test accuracy: {test_score:.1%}")
```
